```python
from fredapi import Fred
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

plt.style.use('fivethirtyeight')
pd.set_option('display.max_columns', 25)
color_pal = plt.rcParams['axes.prop_cycle'].by_key()['color']
```

```python
# Create Fred Object

fred = Fred(api_key = fred_key)
```

```python
# Economic Data

sp_search = fred.search('S&P', order_by='popularity', limit=1000)
sp_search
```

```python
sp500 = fred.get_series(series_id='SP500')
sp500.plot(figsize=(10, 5), title = 'S&P 500', lw = 2)
```

```python
sp500.describe
```

```python
unemployment_search = fred.search(text='unemployment', order_by='series_id')
unemployment_search
```

```python
unemployment_rate = fred.get_series('UNRATE')

unemployment_rate.plot(figsize=(10, 5), title = 'Unemployment Rate', lw = 2)
```

```python
unemployment_state_search = fred.search('unemployment rate state', filter = ('frequency', 'Monthly'), limit=4999)
unemp_df = unemployment_state_search.copy()
unemp_df = unemp_df[
    (unemp_df['seasonal_adjustment'] == 'Seasonally Adjusted') &
    (unemp_df['units'] == 'Percent')
]
unemp_df = unemp_df[unemp_df['title'].str.contains('Unemployment Rate')]
unemp_df.shape
```

```python
all_results = []

for id in unemp_df.index:
    results = fred.get_series(id)
    results = results.to_frame(name=id)
    all_results.append(results)
```

```python
unemp_df = unemp_df[unemp_df['title'].str.contains('Unemployment Rate')]
all_results = pd.concat(all_results, axis = 1)
all_results
```

```python
all_results = all_results.loc[:, all_results.columns.str.len() == 4]
```

```python
unemp_states = all_results.copy()

unemp_states.isna().sum(axis=1).plot()
```

```python
unemp_states = unemp_states.dropna()
```

```python
px.line(unemp_states, title='Unemployment by State')
```

```python
# Unemployment Rate by State in 2020

unemp_states.loc[unemp_states.index == '2020-05-01'].T.sort_values('2020-05-01') \
    .plot(kind= 'bar', figsize = (10, 5), title = 'Unemployment in May 2020 by State').legend().remove()
```

```python
unemp_df['state'] = unemp_df['title'].str.replace('Unemployment Rate in ', '', regex=False)
```

```python
states = unemp_df['state']
states
```

```python
unemp_states = unemp_states.rename(columns=states)
unemp_states
```

```python
ax = (
    unemp_states.loc['2020-05-01']
    .sort_values()
    .plot(kind='barh', figsize=(10, 10), title='Unemployment in May 2020 by State', width=0.9, edgecolor = 'black')
)

ax.tick_params(axis='y', labelsize=10)
ax.legend().remove()
plt.tight_layout()
```

Participation Rate

```python
part_rate_search = fred.search('participation rate state', filter = ('frequency', 'Monthly'), limit=4999)

part_rate_search = part_rate_search[
    (part_rate_search['seasonal_adjustment'] == 'Seasonally Adjusted') &
    (part_rate_search['units'] == 'Percent')]
part_rate_search[part_rate_search['title'].str.contains('California')]
```

```python
all_results = []

part_rate_search = part_rate_search[part_rate_search['title'].str.contains('Participation Rate')]
part_rate_search['state'] = part_rate_search['title'].str.replace('Labor Force Participation Rate for ', '', regex=False)

for id in part_rate_search.index:
    results = fred.get_series(id)
    results = results.to_frame(name=id)
    all_results.append(results)


all_results = pd.concat(all_results, axis = 1)
all_results
```

```python
participation_states = part_rate_search['state']
participation_states
```

```python
all_results = all_results.rename(columns=participation_states)
participation_data = all_results.copy()

cols_to_keep = unemp_states.columns.to_list()

participation_data = participation_data[
    participation_data.columns.intersection(cols_to_keep)
]
```

```python
participation_data
```

```python
ax = (
    participation_data.loc['2020-05-01']
    .sort_values()
    .plot(kind='barh', figsize=(10, 10), title='Participation in May 2020 by State', width=0.9, edgecolor = 'black')
)

ax.tick_params(axis='y', labelsize=10)
ax.legend().remove()
plt.tight_layout()
```
